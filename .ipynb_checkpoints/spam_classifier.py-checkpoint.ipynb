```python
import pandas as pd
message = pd.read_csv('SMSSpamCollection', sep ='\t', names=["label", "message"])
```

```python
message
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
# import re
# import nltk
# from nltk.corpus import stopwords
# from nltk.stem.porter import PorterStemmer

# # Download stopwords (only once)
# nltk.download('stopwords')

# # Initialize stemmer and cache stopwords
# ps = PorterStemmer()
# stop_words = set(stopwords.words('english'))  # Use set for faster lookup

# corpus = []

# for i in range(len(message)):  # message should be your DataFrame
#     # Remove non-alphabetic characters
#     review = re.sub('[^a-zA-Z]', ' ', message['message'][i])
#     review = review.lower()
#     review = review.split()

#     # Stemming and stopword removal
#     review = [ps.stem(word) for word in review if word not in stop_words]

#     # Join processed words back to a single string
#     review = ' '.join(review)
#     corpus.append(review)

corpus = []
for i in range (0, len(message)):
    review = re.sub('[^a-zA-z]', ' ', message['message'][i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

### Create Bag of Word

```python
y=pd.get_dummies(message['label'])
y=y.iloc[:,0].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(corpus, y, test_size=0.2, random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=2500,ngram_range=(1,2))
```

```python
X_train
```

```python
len(X_train), len(y_train)
```

```python
X_train = cv.fit_transform(X_train).toarray()
X_test = cv.transform(X_test).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000, formatter=dict(float=lambda x: "%.3g" % x))
```

### Otput Features

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_detect_model = MultinomialNB().fit(X_train,y_train)
```

```python
y_pred = spam_detect_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
accuracy_score(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

### Creating The TF - IDF Model

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(corpus, y, test_size=0.2, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tv=TfidfVectorizer(max_features= 2500, ngram_range = (1,2))
X_train= tv.fit_transform(X_train).toarray()
X_test = tv.transform(X_test).toarray()
```

```python
X_train
```

```python
tv.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_detect_model = MultinomialNB().fit(X_train,y_train)
```

```python
y_pred = spam_detect_model.predict(X_test)
```

```python
score = accuracy_score(y_test,y_pred)
print(score)
```

```python
print(classification_report(y_test, y_pred))
```
